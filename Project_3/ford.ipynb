```python
X = df.drop(['price'],axis=1)
y = df['price']
```

```python
X
```

```python
y
```

```python
df.columns
```

```python
X_encode = pd.get_dummies(data=X,columns = ['model','transmission','fuelType'],drop_first=True)
```

```python
X_encode
```

```python
X_encode = X_encode.astype(int)
```

```python
X_encode
```

```python
X_encode.columns
```

```python
from sklearn.preprocessing import LabelEncoder
columns = ['model','transmission','fuelType']
```

```python
Xlable = X.copy()
```

```python
label_encoders = {}
```

```python
for col in columns:
    le = LabelEncoder()
    Xlable[col] = le.fit_transform(Xlable[col].astype(str))
    label_encoders[col] = le
```

```python
Xlable
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
numerical_cols = ['year', 'mileage', 'tax', 'mpg', 'engineSize']
scaler = StandardScaler()
X_encode[numerical_cols] = scaler.fit_transform(X_encode[numerical_cols])
```

```python
X_encode
```

```python
Xlable.columns
```

```python
Xlable[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']] = scaler.fit_transform(Xlable[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']])
```

```python
Xlable
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_encode, y, test_size=0.33, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
r2 = r2_score(y_test,y_pred)
r2
```

```python
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_r2 = 1 - ((1 - r2) * (n - 1)) / (n - p - 1)
print("Adjusted R² Score:", adjusted_r2)
```

```python
X_train, X_test, y_train, y_test = train_test_split(Xlable, y, test_size=0.33, random_state=42)
```

```python
model2 = LinearRegression()
model2.fit(X_train,y_train)
```

```python
y_pred = model2.predict(X_test)
```
